# Dataset Label Extractor

This notebook extracts and exports unique labels from various fact-checking datasets.

## Supported Datasets
- **AVeriTeC FEVER** (Train/Dev JSON)
- **FACTors** (CSV)
- **ClaimReview Plus** (HuggingFace)

---

## 1. Setup & Import Libraries

```python
import json
import pandas as pd
from pathlib import Path
from collections import Counter
from datasets import load_dataset
import os


# Set HuggingFace API tokenprint("✓ Libraries imported successfully")

os.environ["HF_TOKEN"] = "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"
```

## 2. Extract Labels from FEVER Train Dataset

```python
# Load FEVER Train dataset
fever_train_path = Path("Datasets/AVeriTeC_FEVER/train.json")

with open(fever_train_path, 'r', encoding='utf-8') as f:
    fever_train_data = json.load(f)

# Extract labels
fever_train_labels = [item.get("label", "") for item in fever_train_data if item.get("label")]
fever_train_label_counts = Counter(fever_train_labels)

print(f"FEVER Train Dataset")
print(f"Total claims: {len(fever_train_data)}")
print(f"Claims with labels: {len(fever_train_labels)}")
print(f"\nUnique labels and counts:")
for label, count in fever_train_label_counts.most_common():
    print(f"  {label}: {count}")

print(f"\nUnique labels list: {sorted(fever_train_label_counts.keys())}")
```

## 3. Extract Labels from FEVER Dev Dataset

```python
# Load FEVER Dev dataset
fever_dev_path = Path("Datasets/AVeriTeC_FEVER/data_dev.json")

with open(fever_dev_path, 'r', encoding='utf-8') as f:
    fever_dev_data = json.load(f)

# Extract labels
fever_dev_labels = [item.get("label", "") for item in fever_dev_data if item.get("label")]
fever_dev_label_counts = Counter(fever_dev_labels)

print(f"FEVER Dev Dataset")
print(f"Total claims: {len(fever_dev_data)}")
print(f"Claims with labels: {len(fever_dev_labels)}")
print(f"\nUnique labels and counts:")
for label, count in fever_dev_label_counts.most_common():
    print(f"  {label}: {count}")

print(f"\nUnique labels list: {sorted(fever_dev_label_counts.keys())}")
```

## 4. Extract Labels from FACTors Dataset

```python
# Load FACTors dataset
factors_path = Path("Datasets/FACTor/FACTors.csv")
factors_df = pd.read_csv(factors_path)

# Extract labels from normalised_rating column
factors_labels = factors_df['normalised_rating'].dropna().astype(str).tolist()
factors_label_counts = Counter(factors_labels)

print(f"FACTors Dataset")
print(f"Total claims: {len(factors_df)}")
print(f"Claims with labels: {len(factors_labels)}")
print(f"\nUnique labels and counts:")
for label, count in factors_label_counts.most_common():
    print(f"  {label}: {count}")

print(f"\nUnique labels list: {sorted(factors_label_counts.keys())}")
```

## 5. Extract Labels from ClaimReview Plus (HuggingFace)

```python
# Load ClaimReview Plus dataset from HuggingFace
print("Loading ClaimReview Plus dataset (this may take a moment)...")
claimreview_dataset = load_dataset("MAI-Lab/ClaimReview2024plus", split="test", token=os.environ.get("HF_TOKEN"))

# Extract labels from rating field
claimreview_labels = [item.get("rating", "") for item in claimreview_dataset if item.get("rating")]
claimreview_label_counts = Counter(claimreview_labels)

print(f"\nClaimReview Plus Dataset")
print(f"Total claims: {len(claimreview_dataset)}")
print(f"Claims with labels: {len(claimreview_labels)}")
print(f"\nUnique labels and counts:")
for label, count in claimreview_label_counts.most_common(20):  # Show top 20
    print(f"  {label}: {count}")

print(f"\nTotal unique labels: {len(claimreview_label_counts)}")
print(f"\nAll unique labels: {sorted(claimreview_label_counts.keys())}")
```

## 6. Summary - All Datasets

```python
# Create summary dictionary
summary = {
    "FEVER_Train": {
        "total_claims": len(fever_train_data),
        "labeled_claims": len(fever_train_labels),
        "unique_labels": sorted(fever_train_label_counts.keys()),
        "label_counts": dict(fever_train_label_counts)
    },
    "FEVER_Dev": {
        "total_claims": len(fever_dev_data),
        "labeled_claims": len(fever_dev_labels),
        "unique_labels": sorted(fever_dev_label_counts.keys()),
        "label_counts": dict(fever_dev_label_counts)
    },
    "FACTors": {
        "total_claims": len(factors_df),
        "labeled_claims": len(factors_labels),
        "unique_labels": sorted(factors_label_counts.keys()),
        "label_counts": dict(factors_label_counts)
    },
    "ClaimReview_Plus": {
        "total_claims": len(claimreview_dataset),
        "labeled_claims": len(claimreview_labels),
        "unique_labels": sorted(claimreview_label_counts.keys()),
        "label_counts": dict(claimreview_label_counts)
    }
}

# Display summary
print("=" * 80)
print("SUMMARY - ALL DATASETS")
print("=" * 80)

for dataset_name, info in summary.items():
    print(f"\n{dataset_name}:")
    print(f"  Total Claims: {info['total_claims']}")
    print(f"  Labeled Claims: {info['labeled_claims']}")
    print(f"  Unique Labels: {len(info['unique_labels'])}")
    print(f"  Labels: {info['unique_labels']}")

print("\n" + "=" * 80)
```

## 7. Export Labels to Files

```python
# Create output directory
output_dir = Path("dataset_labels_export")
output_dir.mkdir(exist_ok=True)

# Export summary to JSON
summary_json_path = output_dir / "all_datasets_labels_summary.json"
with open(summary_json_path, 'w', encoding='utf-8') as f:
    json.dump(summary, f, indent=4, ensure_ascii=False)

print(f"✓ Summary exported to: {summary_json_path}")

# Export individual dataset labels to text files
for dataset_name, info in summary.items():
    # Export unique labels list
    labels_txt_path = output_dir / f"{dataset_name}_unique_labels.txt"
    with open(labels_txt_path, 'w', encoding='utf-8') as f:
        f.write(f"Unique Labels for {dataset_name}\n")
        f.write("=" * 50 + "\n\n")
        for label in info['unique_labels']:
            f.write(f"{label}\n")
    
    print(f"✓ {dataset_name} labels exported to: {labels_txt_path}")
    
    # Export label counts to CSV
    labels_csv_path = output_dir / f"{dataset_name}_label_counts.csv"
    label_df = pd.DataFrame([
        {"label": label, "count": count} 
        for label, count in sorted(info['label_counts'].items(), key=lambda x: -x[1])
    ])
    label_df.to_csv(labels_csv_path, index=False)
    
    print(f"✓ {dataset_name} label counts exported to: {labels_csv_path}")

print(f"\n✓ All exports completed successfully!")
print(f"✓ Output directory: {output_dir.absolute()}")
```
